```python
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
train_df = pd.read_csv("/kaggle/input/competitions/binary-classification-of-student-placement-outcomes/student_placement_train.csv")
test_df = pd.read_csv("/kaggle/input/competitions/binary-classification-of-student-placement-outcomes/student_placement_test.csv")
```

```python
train_df.head()
```

```python
train_df.info()
```

```python
X = train_df.drop(columns=['placed','student_id'])
Y = train_df['placed']
```

```python
num_cols = X.select_dtypes(include=['int','float']).columns
cat_cols = X.select_dtypes(include='object').columns
```

```python
X_train, X_test, y_train,y_test = train_test_split(X,Y,test_size=0.2, stratify=Y ,random_state=42)
```

```python
num_transformer = Pipeline([
    ('imputer',SimpleImputer(strategy='mean')),
    ('num', StandardScaler())    
])
```

```python
cat_transformer = Pipeline([
    ('imputer',SimpleImputer(strategy='most_frequent')),
    ('cat', OneHotEncoder())    
])
```

```python
preprocessor = ColumnTransformer([
    ('num',num_transformer,num_cols),
    ('cat',cat_transformer,cat_cols)
])
```

```python
svm = Pipeline([
    ('preprocessor',preprocessor),
    ('model',SVC())
])
```

```python
svm.fit(X_train,y_train)
```

```python
y_pred = svm.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
print(classification_report(y_test, y_pred))
```

perfomance is bad

lets try randomised search

```python
param_dist = {
    "model__kernel": ["linear", "rbf", "poly"],
    "model__C": [0.01, 1, 10, 100, 1000],
    "model__gamma": ["scale", "auto", 0.001, 0.01, 0.1],
    "model__class_weight": [
        "balanced"
    ],  # Forces all evaluated models to stay balanced!
}
```

```python
random_search = RandomizedSearchCV(
    estimator=svm,
    param_distributions=param_dist,
    n_iter=10,
    cv=3,
    scoring='f1_macro',
    random_state=42
)
```

```python
random_search.fit(X_train, y_train)
```

```python
print(random_search.best_params_)
print(random_search.best_score_)
```

```python
best_model = random_search.best_estimator_

y_pred = best_model.predict(X_test)
```

```python
print("Test Accuracy:", accuracy_score(y_test, y_pred))
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
